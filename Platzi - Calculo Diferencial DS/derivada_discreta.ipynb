```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Definimos la función f(x) = x² — la curva a la que le queremos encontrar la derivada
def f(x):
    return x**2

# Creamos 1000 valores de x entre -10 y 10 para graficar de forma continua
x = np.linspace(-10, 10, num=1000)

# Calculamos la derivada numéricamente usando la definición de límite:
# f'(x) = lim h→0 [f(x+h) - f(x)] / h
# Como h no puede ser exactamente 0, usamos un valor muy pequeño (0.000001)
def df(f):
    h = 0.000001  # h tiende a 0 pero nunca es 0
    return (f(x + h) - f(x)) / h  # Pendiente de la recta secante ≈ recta tangente

y = f(x)  # Evaluamos f(x) en todos los puntos

# Graficamos f(x) en azul y su derivada f'(x) — para f(x) = x², la derivada es f'(x) = 2x
plt.plot(x, y, 'b')   # Curva original f(x) = x²
plt.plot(x, df(f))    # Derivada f'(x) = 2x
plt.grid()
```
