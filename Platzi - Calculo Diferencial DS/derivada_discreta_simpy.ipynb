```python
import numpy as np
from sympy import Symbol, Derivative, lambdify  # Importamos las herramientas de cálculo simbólico de SymPy
import matplotlib.pyplot as plt
```

```python
x = Symbol('x')  # Declaramos 'x' como símbolo matemático — SymPy la tratará como variable algebraica, no como número

# Declaramos la expresión f(x) = x³ + 2x de forma simbólica
expr = x**3 + 2*x

# lambdify convierte la expresión simbólica en una función de Python/NumPy
# Sin esto, no podríamos evaluarla en arreglos de números como np.linspace
f = lambdify(x, expr, 'numpy')

# SymPy calcula la derivada de forma analítica (no numérica como con h→0)
# evaluate=True le indica que resuelva la derivada de inmediato → f'(x) = 3x² + 2
expr_d = Derivative(expr, x, evaluate=True)

# Funcionalizamos la derivada del mismo modo para poder graficarla
f_d = lambdify(x, expr_d)
```

```python
# Generamos 1000 puntos entre -2 y 2 para graficar
X = np.linspace(-2, 2, num=1000)

Y = f(X)    # Evaluamos f(x) = x³ + 2x en todos los puntos
Y_d = f_d(X)  # Evaluamos f'(x) = 3x² + 2 en todos los puntos

# Creamos dos gráficas side-by-side compartiendo el eje Y
fig, ax = plt.subplots(1, 2, sharey=True)
fig.set_size_inches(10, 5)

ax[0].plot(X, Y, color='r')   # Gráfica de f(x) en rojo
ax[1].plot(X, Y_d, color='g') # Gráfica de f'(x) en verde
plt.show()
```
